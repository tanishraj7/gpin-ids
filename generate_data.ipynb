```python
import pandas as pd
import os

# Top 15 features for training
selected_columns = [
    'flow_duration', 'flow_byts_s', 'flow_pkts_s',
    'tot_fwd_pkts', 'tot_bwd_pkts',
    'fwd_pkt_len_mean', 'bwd_pkt_len_mean',
    'pkt_len_var',
    'flow_iat_mean', 'fwd_iat_max', 'bwd_iat_max',
    'pkt_size_avg',
    'syn_flag_cnt', 'ack_flag_cnt',
    'init_fwd_win_byts'
]

# Filepath + labels
data_files = {
    "ddos_suite2_benign_flows.csv": "BENIGN",
    "benign_ddos_suite1_flows.csv": "BENIGN",
    "syn_flood_ddos_flows.csv": "SYN_FLOOD",
    "synonymous_ip_ddos_flows.csv": "IP_SPOOF",
    "tcp_flood_ddos_flows.csv": "TCP_FLOOD",
    "udp_flood_ddos_flows.csv": "UDP_FLOOD",
    "ACK_floods_flows.csv": "ACK_FLOOD",
    "ddos_syn_ack_flows.csv": "SYN_ACK",
    "ddos_udp_frag_flows.csv": "UDP_FRAG",
    "ddos_udpplain.csv": "UDP_PLAIN",
}

# Directory path (change as per Kaggle / Colab mount if needed)
data_dir = "data/"  # Update if needed
combined_df = pd.DataFrame()

# Loop through files and combine
for file, label in data_files.items():
    path = os.path.join(data_dir, file)
    
    # Read only selected cols + first 100k rows
    df = pd.read_csv(path, usecols=selected_columns, nrows=100000)
    df["target"] = label  # add label
    combined_df = pd.concat([combined_df, df], ignore_index=True)

# Save combined output
combined_df.to_csv("gpin_data.csv", index=False)
print("✅ Combined dataset created with shape:", combined_df.shape)
```
